## The route and the rules

### Our route
Ask what extra information changes the available choices. Tower needs the next player as well as the boss position. Array Description needs the neighbouring value as well as the position. Work a small case, describe the stored answer, then write recursion and a table.

### Tower rules · L3-01 to L3-06
Bosses are listed in order: 0 means easy and 1 means hard. Your friend starts. On each turn, the player must defeat the next one or two remaining bosses, then the other player moves. Your friend pays one point for each hard boss they defeat; you pay nothing. Both players cooperate to minimise the friend's total payment.

### Tower names
`n` is the number of bosses. Position `i` identifies the next boss to defeat. `friend_turn` is True when your friend plays next and False when you play next. Position n means no bosses remain; it is just beyond the last valid list index.

### Array rules · L3-07 to L3-13
Fill every missing entry with an integer from 1 to m. A 0 in the input marks a missing entry; 0 is not allowed in the completed array. Keep fixed entries unchanged. Two neighbouring values may differ by at most 1. Count all valid completed arrays, returning the count modulo 1,000,000,007.

### Array names
`n` is the number of entries and `m` is the largest allowed value. `i` is a position, starting at 0; `v` is a possible value placed there. A prefix through position i means the entries from position 0 up to and including position i. For example, the prefix through position 1 of [2, 3, 2] is [2, 3].
